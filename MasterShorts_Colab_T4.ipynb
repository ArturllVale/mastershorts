# MasterShorts - Execucao no Google Colab (GPU T4 + Cloudflare Tunnel)

Este notebook prepara e executa o ecossistema completo do **MasterShorts** com aceleracao por hardware via NVIDIA Tesla T4:
- **Transcribe:** Faster-Whisper rodando na GPU (CUDA float16, modelo `large-v3-turbo`)
- **Encoder:** FFmpeg acelerado via hardware `h264_nvenc`
- **Renderizador:** Remotion Express Service na porta 3100
- **Backend:** FastAPI na porta 8000
- **Frontend:** Interface React Vite na porta 5173
- **Tunel Publico:** Cloudflare Quick Tunnel (`*.trycloudflare.com` gratuito sem necessidade de login)

> **Dica:** Voce pode fazer upload de arquivos de video (.mp4/.mov) diretamente pela interface web sem precisar de cookies do YouTube.

```python
#@title (Opcional) 2. Upload de cookies.txt do YouTube
#@markdown Execute esta celula se quiser colar links do YouTube no Colab para contornar o bloqueio de bot do Google Cloud:
from google.colab import files
import os
print("Selecione o arquivo cookies.txt exportado do navegador:")
uploaded = files.upload()
for fn in uploaded.keys():
    os.makedirs("/content/mastershorts", exist_ok=True)
    dest = "/content/mastershorts/cookies.txt"
    if os.path.exists(dest): os.remove(dest)
    os.rename(fn, dest)
    print(f"cookies.txt instalado em {dest} com sucesso!")
```
